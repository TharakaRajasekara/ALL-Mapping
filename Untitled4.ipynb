```python
import os
import zipfile
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

## 1. Data Collection and Preparation

The dataset was collected from a publicly available Kaggle dataset:
https://www.kaggle.com/datasets/maazmakhdoom/dental-cavity-detection-dataset

The images were categorized into two classes:
- plaque (unhealthy teeth)
- no_plaque (healthy teeth)

A total of 100 images were selected (50 per class) to maintain balance. Images were filtered based on clarity and visibility.

```python
plaque_path = os.path.join(dataset_path, "plaque")
no_plaque_path = os.path.join(dataset_path, "no_plaque")

print("Plaque images:", len(os.listdir(plaque_path)))
print("No plaque images:", len(os.listdir(no_plaque_path)))
```

```python
image_paths = []
labels = []

for img_name in os.listdir(plaque_path):
    image_paths.append(os.path.join(plaque_path, img_name))
    labels.append(1)   # plaque

for img_name in os.listdir(no_plaque_path):
    image_paths.append(os.path.join(no_plaque_path, img_name))
    labels.append(0)   # no_plaque

image_paths = np.array(image_paths)
labels = np.array(labels)

print("Total images:", len(image_paths))
print("Labels:", np.unique(labels, return_counts=True))
```

```python
train_paths, test_paths, train_labels, test_labels = train_test_split(
    image_paths,
    labels,
    test_size=0.2,
    random_state=123,
    stratify=labels
)

print("Training images:", len(train_paths))
print("Testing images:", len(test_paths))
print("Train labels:", np.unique(train_labels, return_counts=True))
print("Test labels:", np.unique(test_labels, return_counts=True))
```

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 16

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.image.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, IMG_SIZE)
    image = image / 255.0
    return image, label
```

## 2. Data Preprocessing

Images were resized to 224×224 pixels and normalized to values between 0 and 1. The dataset was split into training (80%) and testing (20%). Data augmentation techniques such as flipping, rotation, and zoom were applied.

```python
train_data = tf.data.Dataset.from_tensor_slices((train_paths, train_labels))
test_data = tf.data.Dataset.from_tensor_slices((test_paths, test_labels))

train_data = train_data.map(load_image).shuffle(100).batch(BATCH_SIZE)
test_data = test_data.map(load_image).batch(BATCH_SIZE)
```

```python
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])
```

## 3. Model Development

A transfer learning approach using MobileNetV2 was used. This model was chosen because it is efficient and performs well with small datasets.

The model includes:
- Pre-trained MobileNetV2 base
- Global pooling layer
- Dense and dropout layers
- Sigmoid output for binary classification

```python
base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

model = models.Sequential([
    data_augmentation,
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(1, activation="sigmoid")
])
```

## 4. Model Training

The model was trained for 10 epochs using the Adam optimizer and binary cross-entropy loss function.

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    train_data,
    validation_data=test_data,
    epochs=10
)
```

```python
loss, accuracy = model.evaluate(test_data)

print("Test Accuracy:", accuracy)
```

```python
y_true = []
y_pred = []

for images, labels in test_data:
    predictions = model.predict(images)
    predictions = (predictions > 0.5).astype(int)

    y_true.extend(labels.numpy())
    y_pred.extend(predictions.flatten())

y_true = np.array(y_true)
y_pred = np.array(y_pred)
```

## 5. Results and Evaluation

The model achieved the following performance:

- Accuracy: 94%
- Precision: 0.80 (no_plaque), 1.00 (plaque)
- Recall: 1.00 (no_plaque), 0.92 (plaque)
- F1-score: 0.89 (no_plaque), 0.96 (plaque)

The confusion matrix shows strong performance with minimal misclassification.

```python
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)
```

## 7. Model Comparison

A simple CNN model was also tested. MobileNetV2 performed better due to transfer learning, which helps in extracting better features from images.
